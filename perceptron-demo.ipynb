```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/placement-dataset.csv')
```

```python
print(df.shape)
```

```python
sns.scatterplot(x=df['cgpa'],y=df['iq'],hue=df['placement'])
```

```python
x=df.iloc[:,1:3]
y=df.iloc[:,-1]
```

```python
from sklearn.linear_model import Perceptron
p=Perceptron()
```

```python
# Identify rows with NaN values in x
nan_rows_mask = x.isna().any(axis=1)

# Create cleaned versions of x and y by dropping rows with NaNs
x_cleaned = x[~nan_rows_mask]
y_cleaned = y[~nan_rows_mask]

p.fit(x_cleaned, y_cleaned)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x_cleaned.values, y_cleaned.values, clf=p)
```
